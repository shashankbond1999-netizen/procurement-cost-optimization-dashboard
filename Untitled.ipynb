```python
import pandas as pd

df = pd.read_csv('/content/Lead Scoring.csv')

df.shape
df.head()
df.info()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
# Conversion Rate
print("Conversion Rate:", df['Converted'].mean()*100)

# Lead Source Performance
print(df.groupby('Lead Source')['Converted'].mean().sort_values(ascending=False))

# Lead Origin Performance
print(df.groupby('Lead Origin')['Converted'].mean().sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt

df.groupby('Lead Source')['Converted'].mean().sort_values().plot(kind='barh')
plt.title('Conversion Rate by Lead Source')
plt.show()
```

```python
df.to_csv('lead_scoring_cleaned.csv', index=False)
```
